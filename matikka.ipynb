```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

def sigmoid(x):
    return 1/(1 + np.exp(-x))

    
```

```python
# add following values to a numpy array, with one value being binary, and other being continuous
x = np.array([[1.1],
              [4.1],
              [2.7],
              [3.9],
              [7.1],
              [1.3],
              [9.7],
              [0.7],
              [3.4],
              [6.0]])

y = np.array([[0],
              [1],
              [0],
              [0],
              [1],
              [0],
              [1],
              [0],
              [1],
              [0]])

n = x.shape[0]
```

```python
X = np.concatenate((np.ones((n,1)), x), axis=1)

M = np.random.randn(2, 1)
alpha = 0.1

for k in range(10000):
    y_hat = sigmoid(X.dot(M))
    dM = -X.T.dot(y - y_hat)/n
    M = M - alpha*dM
    
z_hat = np.rint(sigmoid(X.dot(M)))

print('Virheiden lkm: ', np.sum(np.abs(z_hat - y)), ' / ', n)
print(M)

plt.figure(figsize=(6, 2))
plt.grid()
plt.plot(x, y, 'bo')
plt.plot(x, z_hat, 'r.')
plt.show()
```
